```python
(ds_train, ds_test), info = tfds.load('fashion_mnist', split=['train', 'test'], with_info=True, as_supervised=True)
```

```python
ds_train



```

```python
# Values before normalization
image_batch, labels_batch = next(iter(ds_train))
print("Before normalization ->", np.min(image_batch[0]), np.max(image_batch[0]))
```

```python
BATCH_SIZE = 32
ds_train_norm= ds_train.map(lambda x, y: (tf.cast(x,tf.float32) / 255.0, y))
ds_test_norm=ds_test.map(lambda x,y : (tf.cast(x,tf.float32) / 255.0,y))
```

```python
#batching

ds_train_batch = ds_train_norm.batch(BATCH_SIZE)
ds_test_batch = ds_test_norm.batch(BATCH_SIZE)
```

```python
# Examine the min and max values of the batch after normalization
image_batch, labels_batch = next(iter(ds_train))
print("After normalization ->", np.min(image_batch[0]), np.max(image_batch[0]))
```

```python
# Define the model
model = tf.keras.models.Sequential([tf.keras.layers.Flatten(),
                                    tf.keras.layers.Dense(64, activation=tf.nn.relu),
                                    tf.keras.layers.Dense(10, activation=tf.nn.softmax)])
```

```python
# Compile the model
model.compile(optimizer = tf.keras.optimizers.Adam(),
              loss = tf.keras.losses.SparseCategoricalCrossentropy(),
              metrics=[tf.keras.metrics.SparseCategoricalAccuracy()])

model.fit(ds_train_batch, epochs=5)
```

```python
model.evaluate(ds_test_batch)
```

```python
#callbacks to stop the training

class mycallback(tf.keras.callbacks.Callback):
  def on_epoch_end(self, epoch, logs={}):
    if(logs.get('sparse_categorical_accuracy')>0.9):
      self.model.stop_training = True
callbacks = mycallback()

model.fit(ds_train_batch, epochs=5, callbacks=[callbacks])
```
